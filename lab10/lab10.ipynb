9. Wypisz dopasowania zawierające datę w formacie YYYY-MM-DD.
10. Wypisz dopasowania zawierające ścieżkę w formacie UNIX (/.../...).
11. Wypisz dopasowania zawierającą adres IP w wersji 4.
12. Wypisz tekst z każdego cytowania (tekst pomiędzy " oraz ").

```python
import re

text = """Adam Malinowski
.gitignore
2023-01-17 error "Page not found"
[2025-03-06] NOTICE "User admin logged in"
Code 3300 was invalid
https://www.onet.pl 200 176353
File /etc/passwd: permission denied
Józef
Ania
JOLA
marek
Kowalski
bodo363
PIN 0000 was invalid
/users/test is not a valid directory name
192.168.0.1 access denied
1000
666"""

print("--- Wyniki zadań ---")


print("Zad 1:", re.findall(r'^[A-ZŻŹĆĄŚĘŁÓŃ].*', text, flags=re.MULTILINE))
print("Zad 2:", re.findall(r'\d+', text))
print("Zad 3:", re.findall(r'^.*\..*$', text, flags=re.MULTILINE))
print("Zad 4:", re.findall(r'\d{3,}', text))
print("Zad 5:", re.findall(r'^.*\d{3,}.*$', text, flags=re.MULTILINE))
print("Zad 6:", re.findall(r'^[A-Za-zŻŹĆĄŚĘŁÓŃżźćńółęąś]+$', text, flags=re.MULTILINE))
print("Zad 7:", re.findall(r'^\d+$', text, flags=re.MULTILINE))
print("Zad 8:", re.findall(r'\b(?:in)?valid\b', text))
print("Zad 9:", re.findall(r'\d{4}-\d{2}-\d{2}', text))
print("Zad 10:", re.findall(r'/(?:[a-zA-Z0-9_-]+/)+[a-zA-Z0-9_-]+', text))
print("Zad 11:", re.findall(r'\b(?:\d{1,3}\.){3}\d{1,3}\b', text))
print("Zad 12:", re.findall(r'"([^"]*)"', text))
print("Zad 13:", re.findall(r'^.{4}$', text, flags=re.MULTILINE))
```
