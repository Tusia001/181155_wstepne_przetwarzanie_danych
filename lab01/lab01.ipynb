Zadanie 3

Na zbiorze danych iris (https://archive.ics.uci.edu/dataset/53/iris) policz (wszystkie cechy - cechy bez typu irysa czyli etykiety):

- globalną wartość średnią dla wszystkich cech,
- globalną medianę dla wszystkich cech,
- globalne odchylenie standardowe dla wszystkich cech,
- wszystkie powyższe wartości dla każdego typu irysów oddzielnie.

```python
import pandas as pd
url_iris = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
iris = pd.read_csv(url_iris, names=['sl', 'sw', 'pl', 'pw', 'class'])
features = iris.drop(columns=['class'])

print(f"Średnia:\n{features.mean()}\n")
print(f"Mediana:\n{features.median()}\n")
print(f"Odchylenie standardowe:\n{features.std()}\n")

print("Statystyki dla każdego typu irysa oddzielnie")
grouped = iris.groupby('class')
print(f"Średnie grupowe:\n{grouped.mean()}\n")
print(f"Mediany grupowe:\n{grouped.median()}\n")
print(f"Odchylenie grupowe:\n{grouped.std()}")
```

Znajdź wartości przedziałów kwartyli dla wszystkich cech zbioru wine quality (oprócz quality oraz color).

Zadanie 6

Wyświetl pozycje ze zbioru wine quality, które znajdują się powyżej 95-go percentyla wartości dla cechy residual_sugar.

```python
url_red = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
url_white = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-white.csv"

red_wine = pd.read_csv(url_red, sep=';')
red_wine['color'] = 'red'

white_wine = pd.read_csv(url_white, sep=';')
white_wine['color'] = 'white'
wine = pd.concat([red_wine, white_wine], ignore_index=True)
wine.columns = wine.columns.str.replace(' ', '_')

print("\n--- ZAD 4")
wariancja_globalna = wine.drop(columns=['quality', 'color']).var().sort_values()
print("Wariancja cech (rosnąco):\n", wariancja_globalna)

wariancja_wg_jakosci = wine.drop(columns=['color']).groupby('quality').var()
print("\nWariancja w grupach jakości:\n", wariancja_wg_jakosci)

srednia_wariancja = wariancja_wg_jakosci.mean(axis=1)
print("\nUśredniona wariancja dla oceny jakości:\n", srednia_wariancja)


odchylenie_wg_koloru = wine.drop(columns=['quality']).groupby('color').std().T
print("\nOdchylenie standardowe (czerwone vs białe):\n", odchylenie_wg_koloru)


print("\n--- ZAD 5")
kwartyle = wine.drop(columns=['quality', 'color']).quantile([0.25, 0.5, 0.75])
print("Przedziały kwartyli:\n", kwartyle)


print("\n--- ZAD 6")
percentyl_95 = wine['residual_sugar'].quantile(0.95)
outliery_cukru = wine[wine['residual_sugar'] > percentyl_95]

print(f"Wartość 95. percentyla dla residual_sugar wynosi: {percentyl_95}")
print(f"Liczba wierszy powyżej tego progu: {len(outliery_cukru)}")
print("Pierwsze 5 pozycji:\n", outliery_cukru.head())
```
