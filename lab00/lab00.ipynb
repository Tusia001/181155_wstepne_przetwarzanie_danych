Zadanie 7
Sprawdź metodą memory_usage() (z parametrem deep=True oraz bez niego) ile danych zajmuje ramka. Wybierz jeden z typów danych numerycznych rodzaju float o mniejszej precyzji i zapisz ramkę w nowej zmiennej iris_smaller, której kolumny numeryczne będą rzutowane na ten nowy typ float. Ponownie sprawdź ile pamięci zostało dla ramki zaalokowane.

```python
mem_standard = iris.memory_usage()
mem_deep = iris.memory_usage(deep=True)

print("Pamięć standardowa:", mem_standard.sum())
print("Pamięć głęboka:", mem_deep.sum())
iris_smaller = iris.copy()
numeric_cols = iris_smaller.select_dtypes(include=['float64']).columns
iris_smaller[numeric_cols] = iris_smaller[numeric_cols].astype('float32')

print("\nPamięć po zmianie na float32:", iris_smaller.memory_usage(deep=True).sum())
```
